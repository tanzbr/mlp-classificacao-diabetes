```python
# === Imports e Configurações ===
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import torch
import torch.nn as nn
from torch.utils.data import TensorDataset, DataLoader

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

print("PyTorch:", torch.__version__)

# Hiperparâmetros
SEED = 42
TEST_SIZE = 0.2
BATCH_SIZE = 32
HIDDEN = 32
DROPOUT = 0.3
EPOCHS = 100
LR = 0.01

np.random.seed(SEED)
torch.manual_seed(SEED)

# Dispositivo
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)
```

## 1) Carregar dataset e inspecionar

```python
# Dataset: Pima Indians Diabetes
# Fonte: https://www.kaggle.com/datasets/uciml/pima-indians-diabetes-database

import os
import zipfile
import urllib.request

# Baixar dataset do Kaggle
url = "https://www.kaggle.com/api/v1/datasets/download/uciml/pima-indians-diabetes-database"
zip_path = "pima-indians-diabetes-database.zip"
csv_filename = "diabetes.csv"

# Download do arquivo zip
print("Baixando dataset...")
urllib.request.urlretrieve(url, zip_path)

# Extrair o zip
print("Extraindo...")
with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall(".")

# Remover o zip após extração
os.remove(zip_path)

# Carregar o CSV
df = pd.read_csv(csv_filename)
print("✓ Dataset carregado!")
print("Dimensão:", df.shape)
print("Colunas:", list(df.columns))
df.head()
```

```python
# Distribuição do target
print("Distribuição de classes (Outcome):")
print(df['Outcome'].value_counts())
print(f"\nProporção: {df['Outcome'].mean()*100:.1f}% diabéticos")
```

## 2) Tratamento de valores ausentes
No dataset Pima, zeros em certas colunas são biologicamente impossíveis (ex: Glucose=0) e representam dados faltantes. Substituímos pela mediana.

```python
# Colunas onde 0 é impossível biologicamente
cols_to_fix = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']

print("Zeros por coluna (antes):")
for col in cols_to_fix:
    zeros = (df[col] == 0).sum()
    print(f"  {col}: {zeros}")

# Substituir 0 pela mediana
for col in cols_to_fix:
    median_val = df[col][df[col] != 0].median()
    df[col] = df[col].replace(0, median_val)

print("\n✓ Zeros substituídos pela mediana.")
```

## 3) Separação e Pré-processamento

```python
# Separar features (X) e target (y)
X = df.drop('Outcome', axis=1).values
y = df['Outcome'].values

# Split treino/teste estratificado
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=TEST_SIZE, stratify=y, random_state=SEED
)

# Normalização (fit apenas no treino)
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train).astype(np.float32)
X_test = scaler.transform(X_test).astype(np.float32)

print("Shapes:", X_train.shape, X_test.shape)
```

## 4) TensorDataset + DataLoader

```python
# Converter para tensores
X_train_t = torch.from_numpy(X_train)
y_train_t = torch.from_numpy(y_train).float()
X_test_t = torch.from_numpy(X_test)
y_test_t = torch.from_numpy(y_test).float()

# Datasets e loaders
train_ds = TensorDataset(X_train_t, y_train_t)
test_ds = TensorDataset(X_test_t, y_test_t)

train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True)
test_loader = DataLoader(test_ds, batch_size=BATCH_SIZE, shuffle=False)

print(f"Batches treino: {len(train_loader)}, teste: {len(test_loader)}")
```
